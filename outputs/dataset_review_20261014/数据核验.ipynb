# 输入指纹核验
只核验四个输入文件的字节和SHA256；不展示原始数据。行数沿用既有核验元数据，不冒充本次全量重算。

```python
from pathlib import Path
import hashlib, json
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p/'o2o_work/config/input_lock.json').exists())
lock = json.loads((root/'o2o_work/config/input_lock.json').read_text())
for spec in lock['files']:
    p = root/spec['path']
    with p.open('rb') as f:
        digest = hashlib.file_digest(f, 'sha256').hexdigest()
    assert digest == spec['sha256'] and p.stat().st_size == spec['bytes']
    print(spec['path'], 'SHA256 PASS', digest)
print('Row counts are prior audited metadata; this cell verifies input bytes and hashes, not a fresh row count.')
```
